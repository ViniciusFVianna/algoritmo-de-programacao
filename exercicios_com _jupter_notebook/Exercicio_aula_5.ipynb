### Escreva um programa que leia um número inteiro e informe se ele é par ou ímpar. Utilize try-except para garantir que o programa não quebre caso o usuário digite texto em vez de números;

```python
try:
    numero = input("Digite um número inteiro: ")
    numero = int(numero)
    if numero % 2 == 0:
        print(f"O número {numero} é par.")
    else:
        print(f"O número {numero} é ímpar.")
except ValueError:
    print("Por favor, digite um número inteiro válido.")
```

### Escreva um programa que leia dois números inteiros e informe qual é o maior, ou se são iguais;


```python
try:
  numero1 = int(input("Digite o primeiro número: "))
  numero2 = int(input("Digite o segundo número: "))

  if(numero1 > numero2):
    print(f"O primeiro número {numero1} é maior que o segundo número {numero2}")
  elif (numero1 == numero2):
    print(f"O primeiro número {numero1} é igual ao segundo número {numero2}")
  else:
    print(f"O segundo número {numero2} é maior que o primeiro número {numero1}")
except:
  print("Os números informados são inválidos")
```

### Escreva um programa que leia uma letra representando uma opção de menu ('A', 'B' ou 'C') e imprima uma mensagem diferente para cada opção. Dica: use o método .upper() na entrada para que o programa aceite tanto letras maiúsculas quanto minúsculas.

```python
def texto_estilizado():
    ESTILOS = {
        "vermelho": "\033[91m",
        "verde": "\033[92m",
        "amarelo": "\033[93m",
        "azul": "\033[94m",
        "magenta": "\033[95m",
        "ciano": "\033[96m",
        "negrito": "\033[1m",
        "sublinhado": "\033[4m",
        "resetar": "\033[0m"
    }

    arte = f"""{ESTILOS['ciano']}{ESTILOS['negrito']}
  _____       _   _
 |  __ \     | | | |
 | |__) |   _| |_| |__   ___  _ __
 |  ___/ | | | __| '_ \ / _ \| '_ \
 | |   | |_| | |_| | | | (_) | | | |
 |_|    \__, |\__|_| |_|\___/|_| |_|
         __/ |
        |___/
{ESTILOS['resetar']}"""

    print(arte)
    print(f"{ESTILOS['verde']}>> Bem-vindo ao mundo do código colorido! <<{ESTILOS['resetar']}\n")

    print(f"{ESTILOS['negrito']}Veja o que dá para fazer nativamente no terminal:{ESTILOS['resetar']}")
    print(f"{ESTILOS['vermelho']}* Este texto é vermelho.{ESTILOS['resetar']}")
    print(f"{ESTILOS['amarelo']}* Este texto é amarelo.{ESTILOS['resetar']}")
    print(f"{ESTILOS['magenta']}{ESTILOS['sublinhado']}* Este texto é magenta e sublinhado!{ESTILOS['resetar']}")
    print(f"{ESTILOS['azul']}{ESTILOS['negrito']}* E este é azul e em negrito.{ESTILOS['resetar']}\n")

    print(f"{ESTILOS['ciano']}Dica: A mágica acontece graças aos códigos de escape ANSI! 🚀{ESTILOS['resetar']}")

def exibir_menu():
    menu = """
=====================================
Olá escolha a opção que deseja fazer!
=====================================
[A] Novo Jogo
[B] Carregar Jogo
[C] Sair
"""
    print(menu)

    opcao = input('Digite a opção desejada: ').strip().upper()

    match opcao:
            case 'A':
                texto_estilizado()
            case 'B':
                print("Carregando jogo...")
            case 'C':
                print("Saindo do jogo...")
            case _:
                print("Opção inválida! Tente A, B ou C.")

if __name__ == "__main__":
    exibir_menu()
```

### Escreva um programa que leia a idade de uma pessoa e informe sua faixa etária usando elif. Aproveite o encadeamento lógico do Python (ex: 13 <= idade <= 17) para as categorias: "Criança" (até 12 anos), "Adolescente" (13 a 17 anos), "Adulto" (18 a 59 anos) ou "Idoso" (60 anos ou mais);

```python
try:
  idade = int(input("Digite a sua idade em (anos): "))

  if idade < 0:
        print("A idade informada é inválida")
  elif idade <= 12:
      print("Essa pessoa é uma Criança")
  elif idade <= 17:
      print("Essa pessoa é um Adolescente")
  elif idade <= 59:
      print("Essa pessoa é um Adulto")
  else:
      print("Essa pessoa é um Idoso")
except ValueError:
    print("Formato informado inválido!!! Digite apenas números inteiros.")
```

### Leia três números inteiros representando os lados de um triângulo. Primeiro, verifique se formam um triângulo válido (a soma de quaisquer dois lados deve ser maior que o terceiro). Se sim, classifique-o. Dica: para testar se é equilátero, você pode usar a sintaxe limpa a == b == c;

```python
try:
  lado1 = int(input("Digite o primeiro lado do triângulo: "))
  lado2 = int(input("Digite o segundo lado do triângulo: "))
  lado3 = int(input("Digite o terceiro lado do triângulo: "))

  if (lado1 + lado2) > lado3 and (lado1 + lado3) > lado2 and (lado2 + lado3) > lado1:
      print("O triângulo é equilátero")
  else:
      print("Não é um triangulo equilátero!")
except ValueError:
  print("Formato informado inválido!!! Digite apenas números inteiros")
```

### Leia o mês (de 1 a 12) e imprima o nome da estação do ano correspondente no hemisfério sul;

```python
try:
  entrada = input("Digite o número do mês (1 a 12): ")

  if not entrada.strip():
    print("Operação cancelada.")
  else:
    if entrada.isdigit():
      month = int(entrada)
    else:
      month = None

    if month in (12, 1, 2):
      season = "Verão"
    elif month in (3, 4, 5):
      season = "Outono"
    elif month in (6, 7, 8):
      season = "Inverno"
    elif month in (9, 10, 11):
      season = "Primavera"
    else:
      season = "Mês inválido"

  if season == "Mês inválido":
    print("Erro: Entrada inválida. Por favor, digite um número válido entre 1 e 12.")
  else:
    print(f"A estação predominante no mês {month} é:\n{season}")

except:
  print("Formato informado inválido!!! Digite apenas números inteiros")
```

### Leia três números inteiros e determine, sem usar laços de repetição ou funções prontas como sort(), qual deles é o do meio (a mediana). Utilize expressões relacionais e encadeamento lógico (ex: a <= b <= c).

```python
try:
  numero1 = int(input("Digite o primeiro número: "))
  numero2 = int(input("Digite o segundo número: "))
  numero3 = int(input("Digite o terceiro número: "))

  if numero1 <= numero2 <= numero3 or numero3 <= numero2 <= numero1:
    print(f"O número {numero2} é o número do meio")
  elif numero2 <= numero1 <= numero3 or numero3 <= numero1 <= numero2:
    print(f"O número {numero1} é o número do meio")
  else:
    print(f"O número {numero3} é o número do meio")
except ValueError:
  print("Formato informado inválido!!! Digite apenas números inteiros")
```

### Leia o dia, mês e ano de uma data. Verifique se ela é válida (trate os meses de 30 e 31 dias, e as regras do mês de fevereiro). Lembre-se: um ano é bissexto se for divisível por 4, exceto os anos divisíveis por 100 que não sejam divisíveis por 400.

```python
try:
  dia = int(input("Digite o dia: "))
  mes = int(input("Digite o mês: "))
  ano = int(input("Digite o ano: "))

  data_valida = False

  if 1 <= mes <= 12 and ano > 0:

      bissexto = (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0)

      if mes == 1 or mes == 3 or mes == 5 or mes == 7 or mes == 8 or mes == 10 or mes == 12:
          if 1 <= dia <= 31:
              data_valida = True

      elif mes == 4 or mes == 6 or mes == 9 or mes == 11:
          if 1 <= dia <= 30:
              data_valida = True

      elif mes == 2:
          if bissexto and 1 <= dia <= 29:
              data_valida = True
          elif not bissexto and 1 <= dia <= 28:
              data_valida = True

  if data_valida:
      print(f"A data {dia:02d}/{mes:02d}/{ano} é VÁLIDA!")
  else:
      print(f"A data {dia:02d}/{mes:02d}/{ano} é INVÁLIDA!")

except ValueError:
    print("Formato inválido! Por favor, digite apenas números inteiros.")
```
