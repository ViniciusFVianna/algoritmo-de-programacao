## Função: `count`

**Descrição:** Gera uma sequência numérica infinita a partir de um valor inicial e com um passo configurável.

**Principais Aplicações:** Numeração de linhas, criação de IDs sequenciais dinâmicos, controle de iterações em loops infinitos.


```python
from itertools import count

# Começa em 10 e pula de 3 em 3
contador = count(10, 3)

for _ in range(5):
    print(next(contador))
```

## Função: `zip-longest`

**Descrição:** Combina iteráveis de tamanhos diferentes, garantindo que nenhum elemento seja descartado e preenchendo as lacunas das sequências menores com um valor padrão (`fillvalue`).

**Principais Aplicações:** Alinhamento de colunas ou tabelas com tamanhos desiguais, mesclagem de dados incompletos.


```python
from itertools import zip_longest

nomes = ['Ana', 'Beto', 'Carla']
notas = [8.5, 9.0]

resultado = list(zip_longest(nomes, notas, fillvalue='Sem Nota'))
print(resultado)
```

## Função: `cycle`

**Descrição:** Repete um iterável indefinidamente em um loop infinito, retornando ao início automaticamente assim que o final é alcançado.

**Principais Aplicações:** Alternância de turnos (Round-Robin), formatação cíclica de cores em relatórios, animações de terminal.


```python
from itertools import cycle, islice

cores = ['Vermelho', 'Verde', 'Azul']
# Usamos islice para limitar a 7 iterações e evitar loop infinito
sequencia = list(islice(cycle(cores), 7))
print(sequencia)
```

## Função: `repeat`

**Descrição:** Repete um único valor infinitamente ou por um número exato de vezes especificado.

**Principais Aplicações:** Fornecer argumentos constantes para funções `map` ou `zip`, preenchimento de parâmetros fixos.


```python
from itertools import repeat

# Repete a string 'OK' 4 vezes
repetidos = list(repeat('OK', 4))
print(repetidos)
```

## Função: `starmap`

**Descrição:** Aplica uma função a argumentos que já estão agrupados em tuplas, desempacotando-os automaticamente (`*`) para passá-los como parâmetros individuais.

**Principais Aplicações:** Processamento de dados tabulares, coordenadas (x, y) ou parâmetros já organizados em listas de tuplas.


```python
from itertools import starmap

# Pares de (base, expoente)
dados = [(2, 3), (3, 2), (5, 2)]
resultados = list(starmap(pow, dados))
print(resultados)
```

## Função: `combinations`

**Descrição:** Gera todas as combinações possíveis de um tamanho fixo a partir de um iterável, onde a ordem dos elementos **não importa** e não há repetição de itens.

**Principais Aplicações:** Formação de comissões, duplas de trabalho, análises estatísticas e loterias.


```python
from itertools import combinations

equipe = ['Ana', 'Beto', 'Carla']
duplas = list(combinations(equipe, 2))
print(duplas)
```

## Função: `permutations`

**Descrição:** Gera todas as permutações (arranjos ordenados) possíveis de um tamanho fixo a partir de um iterável, onde a ordem dos elementos **importa**.

**Principais Aplicações:** Geração de senhas, resolução do problema do caixeiro-viajante, anagramas e testes de ordenação.


```python
from itertools import permutations

letras = ['A', 'B', 'C']
arranjos = list(permutations(letras, 2))
print(arranjos)
```

## Função: `product`

**Descrição:** Calcula o produto cartesiano entre múltiplos iteráveis, funcionando como múltiplos loops `for` aninhados em uma única linha.

**Principais Aplicações:** Busca em grade (Grid Search) para hiperparâmetros de Machine Learning, testes de combinação de atributos.


```python
from itertools import product

tamanhos = ['P', 'M']
cores = ['Azul', 'Preto']
combinacoes = list(product(tamanhos, cores))
print(combinacoes)
```

## Função: `chain`

**Descrição:** Concatena múltiplos iteráveis em uma única sequência contínua, processando-os sequencialmente como se fossem um único grande iterável.

**Principais Aplicações:** Unificação de listas, arquivos ou streams de dados sem gastar memória criando cópias intermediárias.


```python
from itertools import chain

lista1 = [1, 2]
lista2 = [3, 4]
lista3 = [5, 6]

unidos = list(chain(lista1, lista2, lista3))
print(unidos)
```

## Função: `islice`

**Descrição:** Extrai uma fatia (*slice*) de um iterável com base em índices de início, fim e passo, funcionando perfeitamente mesmo com geradores infinitos.

**Principais Aplicações:** Paginação de fluxos de dados, amostragem de dados em tempo real, limitação de loops infinitos.


```python
from itertools import islice, count

# Pega elementos da contagem infinita do índice 5 ao 15 pulando de 2 em 2
fatia = list(islice(count(0), 5, 15, 2))
print(fatia)
```

## Função: `compress`

**Descrição:** Filtra um iterável utilizando uma máscara booleana correspondente, mantendo apenas os elementos onde a máscara for verdadeira (`True`).

**Principais Aplicações:** Filtragem paralela de dados baseada em critérios lógicos ou indicadores binários.


```python
from itertools import compress

dados = ['A', 'B', 'C', 'D', 'E']
mascara = [True, False, True, False, True]
filtrados = list(compress(dados, mascara))
print(filtrados)
```

## Função: `filter false`

**Descrição:** O oposto do `filter`: retorna apenas os elementos de um iterável para os quais a função condicional retorna `False`.

**Principais Aplicações:** Remoção de registros inválidos ou filtragem de itens indesejados baseados em rejeição.


```python
from itertools import filterfalse

numeros = [1, 2, 3, 4, 5, 6]
# Mantém apenas os números ÍMPARES (onde x % 2 == 0 é False)
impares = list(filterfalse(lambda x: x % 2 == 0, numeros))
print(impares)
```

## Função: `drop while`

**Descrição:** Descarta elementos de um iterável enquanto a condição for verdadeira; a partir do momento em que a condição falha, retorna todos os elementos restantes.

**Principais Aplicações:** Ignorar cabeçalhos, comentários iniciais ou dados irrelevantes no início de arquivos e streams.


```python
from itertools import dropwhile

valores = [1, 3, 5, 10, 2, 4]
# Descarta enquanto for menor que 6
resultado = list(dropwhile(lambda x: x < 6, valores))
print(resultado)
```

## Função: `takewihile`

**Descrição:** Mantém e retorna elementos de um iterável enquanto a condição for verdadeira; assim que a condição falha, para imediatamente e descarta o resto.

**Principais Aplicações:** Leitura de dados até atingir um limite, limitação de streams até uma condição de parada.


```python
from itertools import takewhile

valores = [1, 3, 5, 2, 6, 8]
# Pega elementos enquanto forem menores que 6
resultado = list(takewhile(lambda x: x < 6, valores))
print(resultado)
```

## Função: `accumulate`

**Descrição:** Retorna somas acumuladas (ou resultados de outra função binária aplicada progressivamente) ao longo de um iterável.

**Principais Aplicações:** Cálculo de saldos bancários, somas prefixadas (*running totals*) e análises estatísticas cumulativas.


```python
from itertools import accumulate
import operator

numeros = [1, 2, 3, 4, 5]
somas = list(accumulate(numeros))
# Pode usar outras funções, ex: multiplicação
produtos = list(accumulate(numeros, operator.mul))
print('Somas:', somas)
print('Produtos:', produtos)
```

## Função: `groupby`

**Descrição:** Agrupa elementos adjacentes de um iterável que compartilham a mesma chave gerada por uma função de agrupamento.

**Principais Aplicações:** Agrupamento de registros sequenciais por categoria, data ou atributo comum.


```python
from itertools import groupby

dados = 'AABBBCCDA'
grupos = []
for chave, grupo in groupby(dados):
    grupos.append((chave, list(grupo)))
print(grupos)
```

## Função: `tee`

**Descrição:** Cria múltiplos iteradores independentes (geralmente dois) a partir de uma única fonte de dados iterável.

**Principais Aplicações:** Necessidade de iterar sobre o mesmo fluxo de dados em múltiplos locais do código sem esgotar o iterador original.


```python
from itertools import tee

original = [1, 2, 3, 4]
iter1, iter2 = tee(original, 2)

print('Iter 1:', list(iter1))
print('Iter 2:', list(iter2))
```
