```python
def somar(*args):
    somador = 0
    for numero in args:
        somador += numero
    return somador

somar(1, 2, 45, 67, 98)
```

```python
def lista_do_super(**itens):
    for item in itens.items():
        print(item)

lista_do_super(arroz=5, feijao=3, macarrao=2, leite=4)        
```

```python
def is_even(num: int) -> bool:
    """
    Função que determina se um inteiro é par ou não.


    Args:
        num (int): O número inteiro a ser verificado.

    Returns:
        bool: True se o número for par, False caso contrário.

    """
    try:
        if num % 2 == 0: # Testendo se o re'sto da divisão por 2 é igual a 0
            return True
        return False
    except ValueError:
        print("O valor informado não é um número inteiro.")
```

```python
is_even(2.4)
```

```python
def is_palindrome(arg: str) -> bool:
    """Verifica se uma string é um palíndromo.

    A função desconsidera espaços, pontuações e diferenças entre letras maiúsculas 
    e minúsculas durante a verificação.

    Args:
        arg (str): A string de entrada a ser validada como palíndromo.

    Returns:
        bool: Retorna True se a string for um palíndromo e False caso contrário.

    Raises:
        TypeError: Se a entrada fornecida não for do tipo string.
        ValueError: Se a string estiver vazia, contiver apenas espaços ou não possuir 
            nenhum caractere alfanumérico válido.
    """
    if not isinstance(arg, str):
        raise TypeError("A entrada deve ser obrigatoriamente uma string ")

    if not arg or arg.isspace():
        raise ValueError("A entrada não pode ser vazia.")

    valid_characters = [char.lower() for char in arg if char.isalnum()]

    if not valid_characters:
        raise ValueError("A entrada não pode ser vazia.")

    return valid_characters == valid_characters[::-1]
```

```python
assert is_palindrome("Radar") == True
assert is_palindrome("Ame o poema!") == True
assert is_palindrome("Python") == False
assert is_palindrome("A cara rajada da jararaca") == True  
```

```python
print(is_palindrome("Radar"))
print(is_palindrome("Ame o poema!"))
print(is_palindrome("Python"))
print(is_palindrome("A cara rajada da jararaca"))

try:
    print(is_palindrome(12345))
except TypeError as e:
    print(e)

try:
    print(is_palindrome("   "))
except ValueError as e:
    print(e)

try: 
    print(is_palindrome("........-------........"))
except ValueError as e:
    print(e)

try:
    print(is_palindrome("!@#$%^&*()"))
except ValueError as e:
    print(e)    
```
