#### Vamos começar por uma velho problema 
Como repetimos o `input` até o usuário digitar uma entrada válida

```python
while True:
    try:
        ano_nascimento = int(input("Digite o ano de nascimento: "))
    except ValueError:
        print("Por favor, digite um número válido para o ano de nascimento.")
    else:
        break
```

Acima, vimos como repetir um ou mais comandos por um número indefinido de vezes - nesse caso, enquanto uma condição não for satisfeita.

Agora, vamos ver como repetir uma execução por um número definido de vezes.

> Adicionar aspas

```python
alunos: list = ["Alice", "Bob", "Charlie", "David", "Eva"]
vogais = ("a", "e", "i", "o", "u")
for aluno in alunos:
    if aluno.lower().endswith(vogais):
        print(f"Olá, {aluno}!")
```

```python
numeroVezes = 0

while numeroVezes < 5:
    print(f"Esta é a vez número {numeroVezes + 1}")
    numeroVezes += 1
```

Vamos ver algumas formas convenientes de trabalhar com sequência numéricas.

```python
number_list = [1, 2, 3, 4, 5]
for number in number_list:
    print(number)
```
