# Consulta veículo

Exiba ao usuário todas as informações de todos os veículos declarados acima, utilizando um laço para iterar sobre elas.

> [Dica](https://docs.python.org/pt-br/3.13/library/functions.html#zip)

````python
anos = [2023, 2015, 1998]
placas = ["BRA2F19", "GRE1903", "SCI1909"]
modelos = ["Honda Civic", "Ford Fiesta", "Volkswagen Gol"]
tarifas = [399.80, 179.45, 98.90]

for ano, placa, modelo, tarifa in zip(anos, placas, modelos, tarifas):
    carro = {
        "ano": ano,
        "placa": placa,
        "modelo": modelo,
        "tarifa": tarifa
    }
    carro_text = f"""
    Carro:
        Ano: {carro['ano']}
        Placa: {carro['placa']}
        Modelo: {carro['modelo']}
        Tarifa: {carro['tarifa']}
"""
    print(carro_text)
````

# Menu inicial (protótipo)

Mostre ao usuário uma lista de 5 opções, como em uma tela de "Menu Principal". As opções são:

1. Cadastrar veículo
2. Consultar veículo por placa
3. Calcular preço total da frota (aluguel por 1 dia)
4. Veículo mais novo com aluguel abaixo de um valor
5. Sair

Monte um programa em Python que exiba estas opções na tela, peça ao usuário que insira uma opção válida e exiba a opção selecionada (apenas) no seguinte formato:

```python
"O usuário selecionou a opção 3!"
```

Caso o usuário digite uma opção inválida, o programa deve acusar uma exceção, informar ao usuário e pedir que uma nova opção seja inserida. A impressão do Menu só deve ser interrompida se o usuário pressionar a tecla correspondente.

> [Dica](https://docs.python.org/pt-br/3.13/reference/simple_stmts.html#raise)


````python
import subprocess


class Veiculo:

    def __init__(self, placa: str, modelo: str, ano: int, tarifa: float):
        self.placa = placa.upper()
        self.modelo = modelo
        self.ano = ano
        self.tarifa = tarifa


def clear_terminal():
    subprocess.run("clear")


anos = [2023, 2015, 1998]
placas = ["BRA2F19", "GRE1903", "SCI1909"]
modelos = ["Honda Civic", "Ford Fiesta", "Volkswagen Gol"]
tarifas = [399.80, 179.45, 98.90]

frota = [
    Veiculo(placa, modelo, ano, tarifa)
    for placa, modelo, ano, tarifa in zip(placas, modelos, anos, tarifas)
]


def save_vehicle():
    print("--- Cadastrar veículo ---")
    placa = input("Digite a placa do veículo: ")
    ano = int(input("Digite o ano do veículo (Exemplo: 2020): "))
    modelo = input("Digite o modelo do veículo: ")
    tarifa = float(input("Digite a tarifa do veículo: "))

    carro = Veiculo(placa, modelo, ano, tarifa)

    frota.append(carro)
    print(f"Veículo {carro.modelo} cadastrado com sucesso!")


def search_vehicle_by_plate(placa):
    for vehicle in frota:
        if vehicle.placa == placa:
            yield vehicle


def find_all_vehicles():
    for vehicle in frota:
        carro_text = f"""
            Carro:
                Ano: {vehicle.ano}
                Placa: {vehicle.placa}
                Modelo: {vehicle.modelo}
                Tarifa: {vehicle.tarifa}
        """
        print(carro_text)


def calculate_total_fleet_price():
    total_price = sum(vehicle.tarifa for vehicle in frota)
    print(f"Preço total da frota (aluguel por 1 dia): R$ {total_price:.2f}")


def find_newest_vehicle_below_price(max_price):
    max_ano = max(vehicle.ano for vehicle in frota)
    
    found_vehicles = (
        vehicle
        for vehicle in frota
        if vehicle.ano == max_ano and vehicle.tarifa < max_price
    )

    print(f"Veículos mais novo: {max_ano}")
    print(f"Veículos mais novos com aluguel abaixo de R$ {max_price:.2f}:")
    return found_vehicles


def menu():
    while True:
        clear_terminal()
        print(
            f"""
        ======================    
        --- Menu Principal ---
        ======================
        1. Cadastrar veículo
        2. Consultar veículo por placa
        3. Consultar todos os veículos
        4. Calcular preço total da frota (aluguel por 1 dia)
        5. Veículo mais novo com aluguel abaixo de um valor
        6. Sair
        """
        )
        try:
            option = int(input("Escolha uma opção: "))
            match option:
                case 1:
                    save_vehicle()
                case 2:
                    print("--- Consultar veículo por placa ---")
                    search_plate = input(
                        "Digite a placa do veículo que deseja consultar: "
                    )
                    found_vehicles = search_vehicle_by_plate(search_plate)
                    if found_vehicles:
                        for vehicle in found_vehicles:
                            print(vehicle.__dict__)
                    else:
                        print("Veículo não encontrado.")
                case 3:
                    print("--- Consultar todos os veículos ---")
                    find_all_vehicles()
                case 4:
                    print(
                        "--- Calcular preço total da frota (aluguel por 1 dia) ---"
                    )
                    calculate_total_fleet_price()
                case 5:
                    print(
                        "--- Veículo mais novo com aluguel abaixo do valor indicado ---"
                    )
                    max_price = float(
                        input("Digite o valor máximo do aluguel: ")
                    )
                    found_vehicles = find_newest_vehicle_below_price(max_price)

                    for vehicle in found_vehicles:
                        print(f"Veículo : {vehicle.__dict__}")
                case 6:
                    print("Saindo do programa...")
                    break
                case _:
                    clear_terminal()
                    raise ValueError(
                        "Opção inválida. Por favor, insira um número entre 1 e 5."
                    )
        except ValueError:
            clear_terminal()
            print("Entrada inválida. Por favor, insira um número.")


if __name__ == "__main__":
    menu()
````
