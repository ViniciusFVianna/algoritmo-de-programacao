# Questão 1

## Contexto

Um **palíndromo** é uma sequência de caracteres cuja leitura da esquerda para a direita é idêntica à leitura da direita para a esquerda. Na prática da computação e do processamento de linguagem natural, entretanto, quase nunca recebemos textos perfeitamente formatados. Frases famosas como "A cara rajada da jararaca" ou "Ame o poema" contêm espaços, pontuações e variações de caixa (maiúsculas/minúsculas) que precisam ser normalizadas antes da verificação.

## ​Objetivo
​
Implementar uma função que valide entradas textuais, trate casos limites e exceções de forma explícita, e determine se a sequência tratada forma um palíndromo explorando as propriedades das strings como sequências iteráveis em Python.

## ​Especificação dos Requisitos

Seu código recebe um valor de entrada e retorna True (se for palíndromo) ou False (caso contrário).

- ​Se o argumento passado não for do tipo `str`, a função deve levantar um `TypeError` com a mensagem: "A entrada deve ser obrigatoriamente uma string ".
- ​Se a string estiver vazia ou contiver apenas espaços em branco, a função deve levantar um `ValueError` com a mensagem: "A entrada não pode ser vazia.".
- Não discrimine maiúsculas de minúsculas;
- Caracteres que não sejam alfanuméricos (espaços, vírgulas, pontos, hífens...) devem ser ignorados;
- Se não houver pelo menos um cara


Entrada | Saída esperada
---|---
`"Radar"` | `True`
`"Ame o poema!"` | `True`
`Python` | `False`
`A cara da rajada da jararaca` | `True`
`""`|`ValueError`
`12345`|`TypeError`

```python
def eh_palimdromo(entrada: str) -> bool:
    if not isinstance(entrada, str):
        raise TypeError("A entrada deve ser obrigatoriamente uma string ")

    if not entrada or entrada.isspace():
        raise ValueError("A entrada não pode ser vazia.")

    caracteres_validos = [char.lower() for char in entrada if char.isalnum()]

    if not caracteres_validos:
        raise ValueError("A entrada não pode ser vazia.")

    return caracteres_validos == caracteres_validos[::-1]

print(eh_palimdromo("Radar"))
print(eh_palimdromo("Ame o poema!"))
print(eh_palimdromo("Python"))
print(eh_palimdromo("A cara da rejada da jararaca"))

try:
    print(eh_palimdromo(12345))
except TypeError as e:
    print(e)

try:
    print(eh_palimdromo("   "))
except ValueError as e:
    print(e)

try: 
    print(eh_palimdromo("........-------........"))
except ValueError as e:
    print(e)

try:
    print(eh_palimdromo("!@#$%^&*()"))
except ValueError as e:
    print(e)    
```

# Questão 2

Refaça o trabalho da G1 usando `dict` para representar os veículos e suas informações.

```python
import subprocess


def clear_terminal():
    subprocess.run("clear")


anos = [2023, 2015, 1998]
placas = ["BRA2F19", "GRE1903", "SCI1909"]
modelos = ["Honda Civic", "Ford Fiesta", "Volkswagen Gol"]
tarifas = [399.80, 179.45, 98.90]

# A frota agora é uma lista de dicionários
frota = [
    {
        "placa": placa.upper(),
        "modelo": modelo,
        "ano": ano,
        "tarifa": tarifa,
    }
    for placa, modelo, ano, tarifa in zip(placas, modelos, anos, tarifas)
]


def save_vehicle():
    print("--- Cadastrar veículo ---")
    placa = input("Digite a placa do veículo: ")
    ano = int(input("Digite o ano do veículo (Exemplo: 2020): "))
    modelo = input("Digite o modelo do veículo: ")
    tarifa = float(input("Digite a tarifa do veículo: "))

    carro = {
        "placa": placa.upper(),
        "modelo": modelo,
        "ano": ano,
        "tarifa": tarifa,
    }

    frota.append(carro)
    print(f"Veículo {carro['modelo']} cadastrado com sucesso!")


def search_vehicle_by_plate(placa):
    for vehicle in frota:
        if vehicle["placa"] == placa.upper():
            yield vehicle


def find_all_vehicles():
    for vehicle in frota:
        carro_text = f"""
            Carro:
                Ano: {vehicle['ano']}
                Placa: {vehicle['placa']}
                Modelo: {vehicle['modelo']}
                Tarifa: {vehicle['tarifa']}
        """
        print(carro_text)


def calculate_total_fleet_price():
    total_price = sum(vehicle["tarifa"] for vehicle in frota)
    print(f"Preço total da frota (aluguel por 1 dia): R$ {total_price:.2f}")


def find_newest_vehicle_below_price(max_price):
    if not frota:
        print("A frota está vazia.")
        return iter([])

    max_ano = max(vehicle["ano"] for vehicle in frota)
    
    found_vehicles = (
        vehicle
        for vehicle in frota
        if vehicle["ano"] == max_ano and vehicle["tarifa"] < max_price
    )

    print(f"Ano do(s) veículo(s) mais novo(s): {max_ano}")
    print(f"Veículos mais novos com aluguel abaixo de R$ {max_price:.2f}:")
    return found_vehicles


def menu():
    while True:
        clear_terminal()
        print(
            """
        ======================    
        --- Menu Principal ---
        ======================
        1. Cadastrar veículo
        2. Consultar veículo por placa
        3. Consultar todos os veículos
        4. Calcular preço total da frota (aluguel por 1 dia)
        5. Veículo mais novo com aluguel abaixo de um valor
        6. Sair
        """
        )
        try:
            option = int(input("Escolha uma opção: "))
            match option:
                case 1:
                    save_vehicle()
                case 2:
                    print("--- Consultar veículo por placa ---")
                    search_plate = input(
                        "Digite a placa do veículo que deseja consultar: "
                    )
                    found_vehicles = list(search_vehicle_by_plate(search_plate))
                    if found_vehicles:
                        for vehicle in found_vehicles:
                            print(vehicle)
                    else:
                        print("Veículo não encontrado.")
                case 3:
                    print("--- Consultar todos os veículos ---")
                    find_all_vehicles()
                case 4:
                    print(
                        "--- Calcular preço total da frota (aluguel por 1 dia) ---"
                    )
                    calculate_total_fleet_price()
                case 5:
                    print(
                        "--- Veículo mais novo com aluguel abaixo do valor indicado ---"
                    )
                    max_price = float(
                        input("Digite o valor máximo do aluguel: ")
                    )
                    found_vehicles = find_newest_vehicle_below_price(max_price)

                    count_found = 0
                    for vehicle in found_vehicles:
                        print(f"Veículo: {vehicle}")
                        count_found += 1
                    
                    if count_found == 0:
                        print("Nenhum veículo encontrado com esses critérios.")
                case 6:
                    print("Saindo do programa...")
                    break
                case _:
                    clear_terminal()
                    raise ValueError(
                        "Opção inválida. Por favor, insira um número entre 1 e 6."
                    )
        except ValueError as e:
            clear_terminal()
            print(f"Entrada inválida: {e}")
        
        input("\nPressione Enter para continuar...")


if __name__ == "__main__":
    menu()
```
