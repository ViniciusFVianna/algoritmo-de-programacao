1. Escreva um programa que exiba os números de 1 a 10 usando um
laço `for`.

```python
for number in range(1, 11):
    print(number)
```

2. Escreva um programa que exiba os números pares de 2 a 20
usando um laço `for`.

```python
for number in range(2, 20+1, 2):
    print(number)
```

```python
for number in range(1, 20+1):
    if number % 2 == 0:
        print(number)
```

3. Escreva um programa que use `while` para exibir os números de 1 a 5.

```python
number = 0
while number <= 4:
    print(f"Esta é a vez número {number + 1}")
    number += 1
```

4. Crie um programa que exiba a soma dos números de 1 a 100
usando `for`. Ao longo da execução do laço, imprima uma mensagem de progresso da soma a cada 25 iterações, apenas exibindo um texto genérico como "novos 25 números adicionados à soma...".

```python
soma = 0

for i in range(1, 101):
  soma += i
  if i % 25 == 0:
    print("novos 25 números adicionados à soma...")

print(f"Soma total de 1 a 100: {soma}")
```

5. Escreva um programa que peça ao usuário um número e exiba a
tabuada desse número usando `for`.

```python
number = int(input("Digite um número: "))

for i in range(1, 11):
    produto = number * i
    print(f"{number} x {i} = {produto}")
```

6. Escreva um programa que peça ao usuário um número e exiba
todos os números de 1 até o número fornecido usando `while`.

```python
number = int(input("Digite um número: "))
soma = 1
while soma <= number:
    print(f"Esta é a vez número {soma}")
    soma += 1
  
```

7. Escreva um programa que exiba a soma dos números pares de 1 a 50.

```python
soma = sum(range(2, 51, 2))
print(f"A soma dos números pares de 1 a 50 é: {soma}")
  
```

```python
soma = 0

for i in range(1, 51):
    if i % 2 == 0:
        soma += i

print(soma)
```

8. Escreva um programa que conte quantos números de 1 a 100 são
múltiplos de 7.

```python
counter = 0

for number in range(1, 101):
    if number % 7 == 0:
        counter += 1

print(f"Existem {counter} números de 1 a 100 que são múltiplos de 7.")
```

9. Crie uma lista com os números de 1 a 10 e exiba cada elemento usando um laço `for`.

10. Crie uma lista de 5 nomes e exiba-os um por linha usando `for`. Se o nome for mais extenso do que 10 caracteres, imprima uma mensagem genérica de alerta.

```python
names = ["Alice", "Bob", "Charlieeeeeeeeeee", "David", "Eve"]
for name in names:
    if len(name) > 10:
        print(f'Nome muito grande: {name}')
    else:
        print(name)
```
