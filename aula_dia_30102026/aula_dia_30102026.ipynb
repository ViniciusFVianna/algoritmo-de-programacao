# Arquivos
- Um arquivo é uma sequência de bytes registrada de maneira permanete;
- Permitem a persistência de informaçòes de um programa;
- Arquivos são armazenados em dispositivos de memória auxiliar como:
    - Discos(rígidos, sólidos);
    - Pendrive;
    - Cartões SD;

```python
from pathlib import Path
```

```python
sample_data = Path("sample_data")
```

```python
for path in sample_data.iterdir():
    print(path.read_text())
```
