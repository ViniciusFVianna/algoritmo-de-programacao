### Estruturas Condicionais
```python
if condição:
    # Esse bloco roda se a condição for verdadeira
else:
    # Senão, roda esse bloco

if ph > 7:
    print("Básico")
elif ph == 7:
    print("Neutro")
else:
    print("Ácido")
```

---

### Tratamento de Exceções
```python
try:
    int(input('Valor a ser digitado: '))
except ValueError:
    print('Erro de conversão de tipo')
else:
    print('Executado sem o erro mapeado')
finally:
    # Sempre será executado, independente de qualquer coisa
    print('Fim da execução')
```
